```python
class Karyawan:
    def __init__(self, nama, gaji):
        self.nama = nama
        self.gaji = gaji
        self.energi = 100

    def lembur(self):
        print(f"{self.nama} sedang lembur!")
        self.energi -= 30
        self.gaji += 200000
        print(f"sisa energi: {self.energi} dan gaji terbaru adalah {self.gaji}")
karyawan1= Karyawan("Siti", 500000)
karyawan1.lembur()
karyawan2= Karyawan("Andi", 400000)
karyawan2.lembur()
karyawan3= Karyawan("Budi", 300000)
karyawan3.lembur()
karyawan4= Karyawan("Citra", 500000)
karyawan4.lembur()
karyawan5= Karyawan("Dita", 600000)
karyawan5.lembur() 
```

```python
class RekeningBank:
    def __init__(self, nama_pemilik, saldo=0):
        self.nama_pemilik = nama_pemilik
        self.saldo = saldo

    def menabung(self, nominal):
        self.saldo += nominal
        print(f"Berhasil menabung Rp{nominal}. Saldo sekarang: Rp{self.saldo}")

    def tarik_tunai(self, nominal):
        if self.saldo >= nominal:
            self.saldo -= nominal
            print(f"Berhasil menarik Rp{nominal}. Saldo sekarang: Rp{self.saldo}")
        else:
            print("Maaf, saldo tidak mencukupi.")

rekening_saya = RekeningBank("Lifia", 150000)
rekening_saya.tarik_tunai(50000)
rekening_saya.tarik_tunai(200000)
rekening_saya.menabung(300000)
```
